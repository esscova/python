### TENDÊNCIAS DO MOMENTO (RSS)

```python
import pandas as pd
from trendspyg import download_google_trends_rss
from pprint import pprint
```

```python
trends = download_google_trends_rss(geo='BR')

print(f'Encontrado: {len(trends)} trending topics')
print(f'Título primeiro tópico: {trends[0]['trend']}')
```

```python
for trend in trends[:3]:
    print(f"{trend['trend']} - {trend['traffic']}")
    if trend['news_articles']:
        print(f"  {trend['news_articles'][0]['headline']}")    
```

**outputs**

```python
# csv
trends = download_google_trends_rss(geo='BR',
                                    output_format='csv')
trends[:100]
```

```python
# json
trends = download_google_trends_rss(geo='BR',
                                    output_format='json')
pprint(trends)
```

```python
# dataframe
trends = download_google_trends_rss(geo='BR',
                                    output_format='dataframe')
trends
```

### INTERESSE AO LONGO DO TEMPO

```python
from trendspyg import download_google_trends_interest_over_time

dados = download_google_trends_interest_over_time(
    keyword="bitcoin",
    geo="BR",
    timeframe="today 12-m"
)

print(dados[:5])
```

```python
pprint(dados[:5])
```

```python
len(dados)
```

```python
for i in dados:
    print(i['date'], i['value'])
```

```python
# dataframe
dados = download_google_trends_interest_over_time(
    keyword="bitcoin",
    geo="BR",
    timeframe="today 12-m",
    output_format='dataframe'
)
```

```python
dados
```

```python
dados.dtypes
```

```python
dados['date'] = pd.to_datetime(dados['date'])
dados.dtypes
```

```python
dados.head(3)
```

```python
dados.date.dt.date
```

```python
dados['date'] = dados.date.dt.date
dados.head(3)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(dados['date'], dados['value'])
plt.show()
```
